```python
import re
```

```python
text = """
Please contact us at ankita@gmail.com or support@linkific.com.
For technical queries, email admin@example.org.
"""

emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)

print("Extracted Email Addresses:")
print(emails)
```

```python
text = """
Call us at 9876543210 or 9123456789.
"""

phones = re.findall(r'\b\d{10}\b', text)

print(phones)
```

```python
text = """
Important dates are 07-08-2026, 15/08/2026 and 20.08.2026.
"""

dates = re.findall(r'\b\d{2}[-/.]\d{2}[-/.]\d{4}\b', text)

print(dates)
```

```python
text = "Hello!!! Welcome @ Python #Programming."

clean_text = re.sub(r'[^A-Za-z0-9\s]', '', text)

print(clean_text)
```

```python
text = "Hello     Ankita      Welcome    to Python."

clean_text = re.sub(r'\s+', ' ', text).strip()

print(clean_text)
```

```python
text = "Learning #Python and #DataScience with #AI."

hashtags = re.findall(r'#\w+', text)

print(hashtags)
```

```python
text = "Thanks @ankita and @rahul for joining."

mentions = re.findall(r'@\w+', text)

print(mentions)
```

```python
text = """
Visit https://github.com or https://www.google.com
"""

urls = re.findall(r'https?://\S+', text)

print(urls)
```

```python
text = "Python is a Programming Language."

words = re.findall(r'\b[A-Z][a-z]*\b', text)

print(words)
```

```python
email = "ankita@gmail.com"

pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'

if re.match(pattern, email):
    print("Valid email")
else:
    print("Invalid email")
```

```python
text = """
   Hello!!!    This is   a messy TEXT...
   Welcome @@@ to Python!!!
"""

text = re.sub(r'[^A-Za-z0-9\s]', '', text)
text = re.sub(r'\s+', ' ', text).strip()
text = text.lower()

print(text)
```
