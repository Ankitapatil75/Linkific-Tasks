```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
print("Seaborn version:", sns.__version__)
df = sns.load_dataset("titanic")

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.info()
```

```python
sns.countplot(data=df, x="class")

plt.title("Number of Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
sns.barplot(data=df, x="class", y="fare")

plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")

plt.show()
```

```python
sns.boxplot(data=df, x="class", y="age")

plt.title("Age Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Age")

plt.show()
```

```python
sns.violinplot(data=df, x="class", y="age")

plt.title("Age Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Age")

plt.show()
```

```python
sns.histplot(data=df, x="age", kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
sns.scatterplot(data=df, x="age", y="fare", hue="class")

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
pair_data = df[["age", "fare", "pclass", "survived"]].dropna()

sns.pairplot(pair_data, hue="survived")

plt.show()
```

```python
sns.set_theme(style="whitegrid")

sns.boxplot(data=df, x="class", y="fare")

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
sns.set_theme(style="darkgrid")
plt.show()
```
